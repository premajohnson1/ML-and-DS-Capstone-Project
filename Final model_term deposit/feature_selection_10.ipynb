```python

import pandas as pd
from sklearn.model_selection import train_test_split
import time
import numpy as np

from sklearn.preprocessing import StandardScaler
from sklearn.feature_selection import SelectKBest, f_classif
from sklearn.linear_model import LogisticRegression
import pickle
import matplotlib.pyplot as plt


# 1. Train-Test Split and Feature Scaling

def split_scalar(indep_X, dep_Y):

    X_train, X_test, y_train, y_test = train_test_split(
        indep_X,
        dep_Y,
        test_size=0.25,
        random_state=0,
        stratify=dep_Y
    )

    sc = StandardScaler()

    X_train = sc.fit_transform(X_train)
    X_test = sc.transform(X_test)

    return X_train, X_test, y_train, y_test, sc


# 2. Feature Selection using SelectKBest

def selectkbest(X_train, X_test, y_train, n):

    test = SelectKBest(score_func=f_classif, k=n)

    X_train = test.fit_transform(X_train, y_train)

    X_test = test.transform(X_test)

    return X_train, X_test, test


# 3. Confusion Matrix and Model Evaluation

def cm_prediction(classifier, X_test, y_test):

    y_pred = classifier.predict(X_test)

    from sklearn.metrics import confusion_matrix
    from sklearn.metrics import accuracy_score
    from sklearn.metrics import classification_report

    cm = confusion_matrix(y_test, y_pred)

    Accuracy = accuracy_score(y_test, y_pred)

    report = classification_report(y_test, y_pred)

    return classifier, Accuracy, report, X_test, y_test, cm


# 4. Logistic Regression

def logistic(X_train,y_train,X_test,y_test):       
        # Fitting K-NN to the Training set
        from sklearn.model_selection import GridSearchCV
        from sklearn.linear_model import LogisticRegression
        param_grid = {
        'C': [0.01, 0.1, 1, 10, 100],
        'solver': ['lbfgs', 'liblinear'],
        'penalty': ['l2'],
        'class_weight': [None, 'balanced']
    }
        classifier =  GridSearchCV(
    LogisticRegression(max_iter=2000),
    param_grid,
    cv=5,
    refit=True,
    verbose=3,
    n_jobs=-1,
    scoring='f1_weighted'
)   
        classifier.fit(X_train, y_train)
        classifier,Accuracy,report,X_test,y_test,cm=cm_prediction(classifier,X_test,y_test)
        return  classifier,Accuracy,report,X_test,y_test,cm 


def svm_linear(X_train, y_train, X_test, y_test):

    from sklearn.svm import SVC
    from sklearn.model_selection import GridSearchCV

    param_grid = {
        'C': [0.1, 1, 10, 100],
        'kernel': ['linear'],
        'class_weight': [None, 'balanced']
    }

    classifier = GridSearchCV(
        SVC(random_state=42),
        param_grid,
        cv=5,
        refit=True,
        verbose=3,
        n_jobs=-1,
        scoring='f1_weighted'
    )

    classifier.fit(X_train, y_train)

    print("Best Parameters:", classifier.best_params_)

    return cm_prediction(classifier, X_test, y_test)

def svm_NL(X_train, y_train, X_test, y_test):

    from sklearn.svm import SVC
    from sklearn.model_selection import GridSearchCV

    param_grid = {
        'C': [0.1, 1, 10, 100],
        'kernel': ['rbf'],
        'gamma': ['scale', 'auto', 0.01, 0.1, 1],
        'class_weight': [None, 'balanced']
    }

    classifier = GridSearchCV(
        SVC(random_state=42),
        param_grid,
        cv=5,
        refit=True,
        verbose=3,
        n_jobs=-1,
        scoring='f1_weighted'
    )

    classifier.fit(X_train, y_train)

    print("Best Parameters:", classifier.best_params_)

    return cm_prediction(classifier, X_test, y_test)
   
def Navie(X_train, y_train, X_test, y_test):

    from sklearn.naive_bayes import GaussianNB
    from sklearn.model_selection import GridSearchCV

    param_grid = {
        'var_smoothing': np.logspace(-12, -6, 7)
    }

    classifier = GridSearchCV(
        GaussianNB(),
        param_grid,
        cv=5,
        refit=True,
        verbose=3,
        n_jobs=-1,
        scoring='f1_weighted'
    )

    classifier.fit(X_train, y_train)

    print("Best Parameters:", classifier.best_params_)

    return cm_prediction(classifier, X_test, y_test)       
    
def knn(X_train, y_train, X_test, y_test):

    from sklearn.neighbors import KNeighborsClassifier
    from sklearn.model_selection import GridSearchCV

    param_grid = {
        'n_neighbors': [3, 5, 7, 9, 11],
        'weights': ['uniform', 'distance'],
        'metric': ['euclidean', 'manhattan', 'minkowski'],
        'p': [1, 2]
    }

    classifier = GridSearchCV(
        KNeighborsClassifier(),
        param_grid,
        cv=5,
        refit=True,
        verbose=3,
        n_jobs=-1,
        scoring='f1_weighted'
    )

    classifier.fit(X_train, y_train)

    print("Best Parameters:", classifier.best_params_)

    return cm_prediction(classifier, X_test, y_test)
         

def Decision(X_train,y_train,X_test,y_test):       
        # Fitting K-NN to the Training set
        from sklearn.tree import DecisionTreeClassifier
        from sklearn.model_selection import GridSearchCV

    
        param_grid = {
        'criterion': ['gini', 'entropy'],
    'max_depth': [3, 5, 7, 10, 12, None],
    'min_samples_split': [2, 5, 10, 20],
    'min_samples_leaf': [1, 2, 4, 8],
    'max_features': [None, 'sqrt', 'log2']
    }

        classifier = GridSearchCV(
    DecisionTreeClassifier(random_state=42),
    param_grid,
    cv=5,
    refit=True,
    verbose=3,
    n_jobs=-1,
    scoring='f1_weighted'
)
   
        classifier.fit(X_train, y_train)
        classifier,Accuracy,report,X_test,y_test,cm=cm_prediction(classifier,X_test,y_test)
        return  classifier,Accuracy,report,X_test,y_test,cm 

def random(X_train, y_train, X_test, y_test):

    from sklearn.ensemble import RandomForestClassifier
    from sklearn.model_selection import GridSearchCV

    param_grid = {
        'n_estimators': [50, 100, 200],
        'criterion': ['gini', 'entropy'],
        'max_depth': [None, 5, 10, 15],
        'min_samples_split': [2, 5, 10],
        'min_samples_leaf': [1, 2, 4],
        'class_weight': [None, 'balanced']
    }

    classifier = GridSearchCV(
        RandomForestClassifier(random_state=42),
        param_grid,
        cv=5,
        refit=True,
        verbose=3,
        n_jobs=-1,
        scoring='f1_weighted'
    )

    classifier.fit(X_train, y_train)

    print("Best Parameters:", classifier.best_params_)

    return cm_prediction(classifier, X_test, y_test)
```

```python
# Step 1: Split dataset

X_train, X_test, y_train, y_test, sc = split_scalar(
    independent, dependent
)

# Step 2: Feature Selection

X_train, X_test, selector = selectkbest(
    X_train, X_test, y_train, 10
)

# Step 3: Train Logistic Regression

Logistic_classifier, Accuracy, report, X_test, y_test, cm = logistic(
    X_train, y_train, X_test, y_test
)

print("Accuracy:", Accuracy)
print(report)
print(cm)

Svm_linear_classifier, Accuracy, report, X_test, y_test, cm =svm_linear(
    X_train, y_train, X_test, y_test
)

print("Accuracy:", Accuracy)
print(report)
print(cm)

Svm_NL_classifier, Accuracy, report, X_test, y_test, cm =svm_NL(
    X_train, y_train, X_test, y_test
)

print("Accuracy:", Accuracy)
print(report)
print(cm)

Navie_classifier, Accuracy, report, X_test, y_test, cm =Navie(
    X_train, y_train, X_test, y_test
)

print("Accuracy:", Accuracy)
print(report)
print(cm)

knn_classifier, Accuracy, report, X_test, y_test, cm =knn(
    X_train, y_train, X_test, y_test
)

print("Accuracy:", Accuracy)
print(report)
print(cm)

```
