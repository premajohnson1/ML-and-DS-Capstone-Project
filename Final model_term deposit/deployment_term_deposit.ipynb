```python

import pandas as pd
import pickle

# Step 1: Load the saved model
filename = "finalized_random_forest_term_deposit.sav"

loaded_data = pickle.load(open(filename, 'rb'))

classifier = loaded_data['model']
sc = loaded_data['scaler']
features = loaded_data['features']

print("Random Forest Model Loaded Successfully!")

# Step 2: Get customer input
new_customer = pd.DataFrame({
    'Housing_Loan': [int(input("Enter Housing Loan (0/1): "))],

    'Duration': [float(input("Enter Duration: "))],

    'Days_Since_Previous_Contact': [
        int(input("Enter Days Since Previous Contact: "))
    ],

    'Previous_Campaign_Contacts': [
        int(input("Enter Previous Campaign Contacts: "))
    ],

    'Contact_unknown': [
        int(input("Enter Contact Unknown (0/1): "))
    ],

    'Last_Contact_Month_mar': [
        int(input("Enter Month March (0/1): "))
    ],

    'Last_Contact_Month_may': [
        int(input("Enter Month May (0/1): "))
    ],

    'Last_Contact_Month_oct': [
        int(input("Enter Month October (0/1): "))
    ],

    'Previous_Campaign_Outcome_success': [
        int(input("Enter Previous Outcome Success (0/1): "))
    ],

    'Previous_Campaign_Outcome_unknown': [
        int(input("Enter Previous Outcome Unknown (0/1): "))
    ]
})

# Step 3: Arrange columns in training order
new_customer = new_customer[features]

# Step 4: Scale the customer input
new_customer_scaled = sc.transform(new_customer)

# Step 5: Predict the output
prediction = classifier.predict(new_customer_scaled)

print("\nPrediction:", prediction)

# Step 6: Display the prediction result
if prediction[0] == 1:
    print("Customer is likely to Subscribe")
else:
    print("Customer is unlikely to Subscribe")

```
