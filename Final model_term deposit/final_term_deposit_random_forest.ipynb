```python
dataset.rename(columns={
    'age':'Age',
    'job':'Job',
    'marital':'Marital',
    'education':'Education',
    'default':'Default',
    'balance':'Balance',
    'housing':'Housing_Loan',
    'loan':'Personal_Loan',
    'contact':'Contact',
    'day': 'Last_Contact_Day',
    'month': 'Last_Contact_Month',
    'duration':'Duration',
    'campaign': 'Current_Campaign_Contacts',
    'pdays': 'Days_Since_Previous_Contact',
    'previous': 'Previous_Campaign_Contacts',
    'poutcome': 'Previous_Campaign_Outcome',
    'deposit':'Subscribe'
},inplace=True)
dataset
```

```python
dataset['Default'] = dataset['Default'].map({'yes': 1, 'no': 0})
dataset['Housing_Loan'] = dataset['Housing_Loan'].map({'yes': 1, 'no': 0})
dataset['Personal_Loan'] = dataset['Personal_Loan'].map({'yes': 1, 'no': 0})
dataset['Subscribe']=dataset['Subscribe'].map({'yes': 1, 'no': 0})
dataset
```

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()

dataset['Education'] = le.fit_transform(dataset['Education'])

dict(zip(le.classes_, le.transform(le.classes_)))
```

```python


dataset
```

```python
dataset=pd.get_dummies(dataset,drop_first=True,dtype=int)
dataset
```

```python
dataset.isnull().sum()
```

```python
independent = dataset[[
    'Housing_Loan',
    'Duration',
    'Days_Since_Previous_Contact',
    'Previous_Campaign_Contacts',
    'Contact_unknown',
    'Last_Contact_Month_mar',
    'Last_Contact_Month_may',
    'Last_Contact_Month_oct',
    'Previous_Campaign_Outcome_success',
    'Previous_Campaign_Outcome_unknown'
]]

dependent = dataset['Subscribe']
```

```python
independent
```

```python
dependent
```

```python
#split into training set and test
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(independent, dependent, test_size = 0.25, random_state = 42)
```

```python
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
X_train= sc.fit_transform(X_train)
X_test= sc.transform(X_test)
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import GridSearchCV

param_grid = {
        'n_estimators': [50, 100, 200],
        'criterion': ['gini', 'entropy'],
        'max_depth': [None, 5, 10, 15],
        'min_samples_split': [2, 5, 10],
        'min_samples_leaf': [1, 2, 4],
        'class_weight': [None, 'balanced']
    }

classifier = GridSearchCV(
        RandomForestClassifier(random_state=42),
        param_grid,
        cv=5,
        refit=True,
        verbose=3,
        n_jobs=-1,
        scoring='f1_weighted'
    )

classifier.fit(X_train, y_train)

print("Best Parameters:", classifier.best_params_)

```

```python
grid_predictions = classifier.predict(X_test) 


from sklearn.metrics import confusion_matrix
cm = confusion_matrix(y_test, grid_predictions)
print("The confusion Matrix:\n",cm)

# print classification report 
from sklearn.metrics import classification_report
clf_report = classification_report(y_test, grid_predictions)
print("The classification report:\n",clf_report)
```
